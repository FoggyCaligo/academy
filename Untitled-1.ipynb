```python
class MovieTicket:
    def __init__(self, title, price):
        self.title = title
        self.price = price
    #할인 적용 메소드
    def apply_discount(self, discount):
        self.price -= discount
        print(f"할인 적용 후 가격: {self.price}")

ticket = MovieTicket("avengers", 12.00)
ticket.apply_discount(2.00)
```
